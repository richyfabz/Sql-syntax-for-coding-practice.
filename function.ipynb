# SQL functions

This notebook was empty. These examples use the shared PostgreSQL sample data.

```sql
-- Clean text for reporting without changing raw_customer.
SELECT customer_id, NULLIF(LOWER(TRIM(email)), '') AS email,
    NULLIF(UPPER(TRIM(country)), '') AS country,
    NULLIF(TRIM(customer_name), '') AS customer_name
FROM raw_customer
ORDER BY customer_id;
```
